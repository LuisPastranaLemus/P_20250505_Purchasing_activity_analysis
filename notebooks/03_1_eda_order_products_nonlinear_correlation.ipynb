# 🛒 Instacart EDA: Product Count vs Days Since Prior Order


In this notebook, we explore whether there's a correlation between the **number of products ordered** in a purchase and the **days until the next order** in the Instacart dataset.

We'll examine linear and non-linear relationships using Pearson, Spearman, and Kendall correlations, and visualize the data using an interactive scatter plot.


```python

import pandas as pd
import plotly.express as px
from scipy.stats import pearsonr, spearmanr, kendalltau
```

```python
# Load merged data with number of products and days since prior order
df = pd.read_csv("../data/processed/product_activity/instacart_products_vs_days.csv")
df.head()
```

```python

# Calculate correlations
pearson_corr, _ = pearsonr(df['num_products'], df['days_since_prior_order'])
spearman_corr, _ = spearmanr(df['num_products'], df['days_since_prior_order'])
kendall_corr, _ = kendalltau(df['num_products'], df['days_since_prior_order'])

# Show results
correlation_df = pd.DataFrame({
    'Correlation Type': ['Pearson', 'Spearman', 'Kendall'],
    'Correlation Value': [pearson_corr, spearman_corr, kendall_corr]
})
correlation_df
```

| Type | What it measures | When to use it |
| ------------ | ---------------------------------------------------------- | ------------------------------------------------------------- |
| **Pearson** | **Linear** relationship between two continuous variables | When the data have a normal distribution and a linear relationship |
| **Spearman** | **Monotonic** relationship (increasing or decreasing, not linear) | When there is **order** but no linear shape |
| **Kendall** | Similar to Spearman, but more robust to outliers | Useful with ordinal data or many repetitions / ties |

| (Absolute) Value | General Interpretation | What does it mean? |
| ---------------- | ------------------------------- | ------------------------------------------------------------------------------- |
| 0.00 – 0.10 | 🔸 **Very Weak / No Relationship** | Virtually no relationship (or too weak to be relied upon) |
| 0.10 – 0.30 | 🔹 **Weak** | Slight relationship exists, but may be sensitive to noise |
| 0.30 – 0.50 | 🟡 **Moderate** | Visible relationship, but not strong |
| 0.50 – 0.70 | 🟠 **Strong** | Good relationship, decisions can be made with more confidence |
| 0.70 – 0.90 | 🔴 **Very Strong** | Strong association between variables |
| 0.90 – 1.00 | 🔥 **Almost perfect / Perfect** | The variables move practically together (or are the same thing in disguise!) |


### 📊 Interpretation:
- **Pearson** detects only weak linear relationship.
- **Spearman** and **Kendall** show a slightly stronger (but still weak) monotonic relationship.
- This suggests that there might be some behavioral pattern, but not strong or necessarily linear.


```python

# Interactive scatter plot with trendline
fig = px.scatter(
    df,
    x='num_products',
    y='days_since_prior_order',
    title='Instacart: Number of Products vs. Days Since Prior Order',
    labels={'num_products': 'Number of Products', 'days_since_prior_order': 'Days Since Prior Order'},
    opacity=0.5,
    trendline='ols'
)
fig.show()
```
